```python
from sklearn.tree import DecisionTreeClassifier
```

### Base risco de credito

```python
import pickle
with open('risco_credito.pkl', 'rb') as f:
    x_risco_credito, y_risco_credito = pickle.load(f)
```

```python
x_risco_credito
```

```python
y_risco_credito
```

```python
arvore_risco_credito = DecisionTreeClassifier(criterion= 'entropy')
arvore_risco_credito.fit(x_risco_credito, y_risco_credito)    
```

```python
arvore_risco_credito.feature_importances_
```

```python
arvore_risco_credito.classes_
```

```python
from sklearn import tree
```

```python
import matplotlib.pyplot as plt
previsores = ['história', 'divida', 'garantias', 'rendas']
figura, eixos = plt.subplots(nrows=1, ncols=1, figsize=(10,10))
tree.plot_tree(arvore_risco_credito, feature_names=previsores, class_names=arvore_risco_credito.classes_, filled = True);
```

```python
previsoes = arvore_risco_credito.predict([[0,0,1,2], [2,0,0,0]])
previsoes
```

### Base credito

```python
with open('credit.pkl', 'rb') as f:
    x_credit_treinamento, y_credit_treinamento, x_credit_teste, y_credit_teste = pickle.load(f)
```

```python
x_credit_treinamento.shape, y_credit_treinamento.shape
```

```python
x_credit_teste.shape, y_credit_teste.shape
```

```python
arvore_credit = DecisionTreeClassifier(criterion='entropy', random_state= 0)
arvore_credit.fit(x_credit_treinamento, y_credit_treinamento)
```

```python
previsoes = arvore_credit.predict(x_credit_teste)
previsoes
```

```python
y_credit_teste
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_credit_teste, previsoes)
```

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(arvore_credit)
cm.fit(x_credit_treinamento, y_credit_treinamento)
cm.score(x_credit_teste, y_credit_teste)
```

## Random forest

```python
from sklearn.ensemble import RandomForestClassifier 
```

```python
with open('credit.pkl', 'rb') as f:
    x_credit_treinamento, y_credit_treinamento, x_credit_teste, y_credit_teste = pickle.load(f)
```

```python
x_credit_treinamento.shape, y_credit_treinamento.shape
```

```python
x_credit_teste.shape, y_credit_teste.shape
```

```python
florestaRandomica = RandomForestClassifier(n_estimators=40,criterion='entropy',random_state=0)
florestaRandomica.fit(x_credit_treinamento, y_credit_treinamento)
```

```python
previsoes = florestaRandomica.predict(x_credit_teste)
previsoes
```

```python
y_credit_teste
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(previsoes, y_credit_teste)
```

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(florestaRandomica)
cm.fit(x_credit_treinamento, y_credit_treinamento)
cm.score(x_credit_teste, y_credit_teste)
```
